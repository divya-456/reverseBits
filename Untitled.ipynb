```python
import pickle
```

```python
class Car:
    def __init__(self,fuel,efficiency):
        self._fuel = fuel
        self._efficiency = efficiency
        
    def drive(self,distance):
        max_distance = self._fuel * self._efficiency

        if distance > max_distance:
            print(f"Travelled {max_distance}, Out of fuel!")
        else:
            self._fuel -= distance/ self._efficiency
            
    @property
    def get_fuel(self):
        return f"{self._fuel}L" 
```

```python
firstCar1 = Car(20,5)
```

```python
firstCar1.drive(50)
```

```python
import pickle
```

```python
with open("fast_car_object.pkl","wb") as out_file:
    pickle.dump(firstCar1,out_file)
```

```python
with open("fast_car_object.pkl","rb") as in_file:
    firstCar2 = pickle.load(in_file)
```

```python
firstCar2.get_fuel
```
