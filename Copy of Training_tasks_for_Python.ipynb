```python
# create naive script here

try:

  num1 = int(input("Enter first number: "))
  num2 = int(input("Enter second number: "))
  operator = input("Enter operator: ")

  if operator == "+":
    print(num1 + num2)
  elif operator == "-":
    print(num1 - num2)
  elif operator == "*":
    print(num1 * num2)
  elif operator == "/":
    print(num1 / num2)

except ValueError:
  print("Invalid input")
except ZeroDivisionError:
  print("Cannot divide by zero")
except Exception as e:
  print(e)
```

```python
# create function based script here

def sum(num1,num2):
  return num1 + num2

def subtract(num1,num2):
  return num1 - num2

def multiply(num1,num2):
  return num1 * num2

def divide(num1,num2):
  return num1 / num2

try:

  num1 = int(input("Enter first number: "))
  num2 = int(input("Enter second number: "))
  operator = input("Enter operator: ")

  if operator == "+":
    print(sum(num1,num2))
  elif operator == "-":
    print(subtract(num1,num2))
  elif operator == "*":
    print(multiply(num1,num2))
  elif operator == "/":
    print(divide(num1,num2))

except ValueError:
  print("Invalid input")
except ZeroDivisionError:
  print("Cannot divide by zero")
except Exception as e:
  print(e)
```

```python
# create class based script here

class Calculator:
  def __init__(self,num1,num2):
    self.num1 = num1
    self.num2 = num2

  def sum(self):
    return self.num1 + self.num2

  def subtract(self):
    return self.num1 - self.num2

  def multiply(self):
    return self.num1 * self.num2

  def divide(self):
    return self.num1 / self.num2

try:

  num1 = int(input("Enter first number: "))
  num2 = int(input("Enter second number: "))
  calc = Calculator(num1,num2)
  operator = input("Enter operator: ")

  if operator == "+":
    print(calc.sum())
  elif operator == "-":
    print(calc.subtract())
  elif operator == "*":
    print(calc.multiply())
  elif operator == "/":
    print(calc.divide())


except ValueError:
  print("Invalid input")
except ZeroDivisionError:
  print("Cannot divide by zero")
except Exception as e:
  print(e)

```

## W.A.P. that calculates result of a student. Student has given exam for five subjects. Passing marks are 23 out of 70


```python
# create naive script here
try:
  name = input("Enter name: ")
  marks = []

  for j in range(5):
    marks.append(int(input("Enter marks: ")))

  total = sum(marks)
  percentage = (total / 350)*70

  if percentage >= 23:
    print(f"{name} - {percentage}% - passed")
  else:
    print(f"{name} - {percentage}% - failed")

except ValueError as e:
  print(e)
except Exception as e:
  print(e)
```

```python
# create function based script here

def result(marks):
  total = sum(marks)
  percentage = (total / 350)*70

  if percentage >= 23:
    print(f"{name} - {percentage}% - passed")
  else:
    print(f"{name} - {percentage}% - failed")

try:
  name = input("Enter name: ")
  marks = []

  for j in range(5):
    marks.append(int(input("Enter marks: ")))

  result(marks)
except ValueError as e:
  print(e)

except Exception as e:
  print(e)
```

```python
# create class based script here
class Result:
  def __init__(self,name,marks):
    self.name = name
    self.marks = marks

  def calc_result(self):
    total = sum(self.marks)
    percentage = (total / 350)*70

    if percentage >= 23:
      print(f"{self.name} - {percentage}% - passed")
    else:
      print(f"{self.name} - {percentage}% - failed")

try:
  name = input("Enter name: ")
  marks = []

  for j in range(5):
    marks.append(int(input("Enter marks: ")))

  result = Result(name,marks)
  result.calc_result()
except ValueError as e:
  print(e)

except Exception as e:
  print(e)
```

```python
# create class based script here

class Stack:
    def __init__(self):
        self.stack=[]
    def push(self,element):
        self.stack+=[element]
        print(f"{element} is added.")
        
    def pop(self):
        try:
            print(f"{self.stack[-1]} is deleted.")
            self.stack[:]=self.stack[:-1]
        except Exception as error:
            print("Stack is empty!",error)
            
    def peek(self):
        try:
            print(f"peek: {self.stack[-1]}")
            
        except Exception as error:
            print("Stack is empty!",error)
            
    def display(self):
        print("Stack=",self.stack)
        
stack_object=Stack()
while True:
    print("1)push")
    print("2)pop")
    print("3)peek")
    print("4)display")
    print("5)exit")
    choice=int(input("Enter your choice: "))
    match choice:
        case 1:
            n=int(input("Enter a number: "))
            stack_object.push(n)
        case 2:stack_object.pop()
        case 3:stack_object.peek()
        case 4:stack_object.display()
        case 5:
            print("Thank You!")
            break
        case _:
            
            print("Invalid Choice!")
            
```

## Sorting algorithms

#### Selection Sort

```python
# test cases : 5 1 3 2 4 - passed
# test cases : 1 2 3 2 3 2 3 - passed
try:
    arr = [int(i) for i in input("Enter an array seperated by spaces: ").split()]
    for i in range(len(arr)):
        print(arr)
        print(arr[i:])
        value = min(arr[i:])
        index_of_value = arr[i:].index(value) + i
        arr[i],arr[index_of_value] = arr[index_of_value],arr[i]
    print(arr)
except ValueError as e:
    print(e)
except IndexError as e:
    print(e)
except Exception as e:
    print(e)
```

#### Bubble Sort

```python
try:
    arr = [int(i) for i in input("Enter an array seperated by spaces: ").split()]
    for i in range(len(arr)):
        for j in range(0,len(arr)-i-1):
            if arr[j]>arr[j+1]:
                arr[j],arr[j+1] = arr[j+1],arr[j]
    print(arr)
except Exception as e:
    print(e)
```

#### Insertion Sort

```python
try:
    arr = [int(i) for i in input("Enter an array seperated by spaces: ").split()]
    for i in range(1,len(arr)):
        insert_index = i
        current_value = arr[i]
        for j in range(i-1, -1, -1):
            if arr[j] > current_value:
                arr[j+1] = arr[j]
                insert_index = j
            else:
                break
        arr[insert_index] = current_value
    print(arr)
except Exception as e:
    print(e)
    
    
```

#### Merge Sort

```python
def merge(left, right):
    result = []
    i = j = 0
    
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
            
    result.extend(left[i:])
    result.extend(right[j:])
    
    return result

def merge_sort(arr):
    step = 1 
    length = len(arr)
    
    while step < length:
        for i in range(0, length, 2 * step):
            left = arr[i:i + step]
            right = arr[i + step:i + 2 * step]
            
            merged = merge(left, right)
            
            for j, val in enumerate(merged):
                arr[i + j] = val
                
        step *= 2
        
    return arr


try:
    arr = [int(i) for i in input("Enter an array seperated by spaces: ").split()]
    sorted_arr = merge_sort(arr)
    print(sorted_arr)

except Exception as e:
    print(e)
```

#### Quick Sort

```python
def partition(array, low, high):
    pivot = array[high]
    i = low - 1
    
    for j in range(low, high):
        if array[j] <= pivot:
            i += 1
            array[i], array[j] = array[j], array[i]

    array[i+1], array[high] = array[high], array[i+1]
    return i+1

def quicksort(array, low=0, high=None):
    if high is None:
        high = len(array) - 1

    if low < high:
        pivot_index = partition(array, low, high)
        quicksort(array, low, pivot_index-1)
        quicksort(array, pivot_index+1, high)

try:
    arr = [int(i) for i in input("Enter an array seperated by spaces: ").split()]
    quicksort(arr)
    print(arr)

except Exception as e:
    print(e)
```

## Searching Algorithms

#### Linear Search

```python
arr = [int(i) for i in input("Enter an array seperated by spaces: ").split()]
value = int(input("Enter the value you want to search: "))
index = -1
for i in range(len(arr)):
    if arr[i] == value:
        index = i
        break
if index == -1:
    print("Value not found!")
else:
    print(f"Your value is at {index}")
```

#### Binary Search

```python
arr = [int(i) for i in input("Enter a sorted array seperated by spaces: ").split()]
value = int(input("Enter the value you want to search: "))
left = 0
right = len(arr) - 1
index = -1
while left <=right:
    mid = (left+right)//2
    if arr[mid] == value:
        index = mid
        break
    else:
        if arr[mid] > value:
            right = mid-1
        else:
            left = mid+1
if index == -1:
    print("Value not found!")
else:
    print(f"Value is at {index}")
```

```python
import random
from prettytable import PrettyTable

class Bank:
    instances = []
    bank_amount = 5000000
    global account_dict
    global user
    def __init__(self):
        self.create_account()
        self.transaction = []
        Bank.instances.append(self)
        self.has_loan = False

    def get_account_dict(self):
        print(account_dict)

    def create_account(self):
        name = input("What is your name? ")
        self.name = name
        self.account_no = random.randrange(1000,5000)
        while True:
            pin = int(input("Create your pin: "))
            if len(str(pin)) == 4:
                self.pin = pin
                break
            else:
                print("Please enter a pin of length 4")
        print("Your account has been created successfully!")
        print(f"Your account details\nAccount number : {self.account_no}\nPin : {self.pin}")
        account_dict[self.account_no] = self

        
        while True:
            print("The initial deposit must be greater than 5000")
            amount = int(input("Please enter the initial amount you would like to deposit : "))
            if amount < 5000:
                print("Please enter an amount greater then 5000")
            else:
                self.amount = amount
                print("Deposit successfull")
                break
        self.is_login = True
    
    
    def login(self):
        print("Hello Sir,")
        if self.is_login:
            print("You are already logged in!")
        else:
            while True:
                account = int(input("Please enter your account number : "))
                if account not in account_dict:
                    print("No such account found!")
                    return

                user = account_dict[account]
                pin = int(input("Please enter your 4-digit pin : "))
    
                if account == user.account_no and pin == user.pin:
                    print("Login Successfull !")
                    user.is_login = True
                    break
                else:
                    print("Please enter valid details!")


    def logout(self):
        self.is_login = False
        print("Logout successfull")

    def deposit(self,amount):
        if self.is_login:
            self.amount += amount
            transaction = {
                "amount" : amount,
                "type" : "credited",
                "from" : self.name,
                "to" : self.name
            }
            self.transaction.append(transaction)
            print("Amount deposited successfully!")
        else:
            print("Please login first")

    def withdraw(self,amount):
        if not self.is_login:
            print("Please login first")
            return
            
        if self.amount - amount < 5000:
           print("Insufficient balance!")
        else:
            self.amount -= amount
            transaction = {
                "amount" : amount,
                "type" : "debited",
                "from" : self.name,
                "to" : self.name
            }
            self.transaction.append(transaction)
            print("Withdraw successfull")


    def close_account(self):
        if self.is_login:
            del self
            print("Account closed successfully!")
        else:
            print("Please login first")

    def transfer_amount(self):
        if not self.is_login:
            print("Please login first")
            return

        account_no = int(input("Enter the account no you want to transfer the amount to : "))
        if account_no not in account_dict:
            print("There is no such account! ")
            return
        other = account_dict[account_no]

        
        amount = int(input("Enter the amount you want to transfer"))
        if self.amount - amount < 5000:
            print("Insufficient balance!")
        else:
            self.amount -= amount
            other.amount += amount
            print("Amount transferred successfully")
            transaction = {
                "amount" : amount,
                "type" : "debited",
                "from" : self.name,
                "to" : other.name
            }
            self.transaction.append(transaction)
            

    def get_transactions(self):
        if not self.is_login:
            print("Please login first")
            return
        if self.transaction:
            t = PrettyTable(self.transaction[0].keys())
            for transaction in self.transaction:
                t.add_row(transaction.values())
            print(t)
            print(f"Your total balance is {self.amount}")
        else:
            print("You have no transactions")
        
    
    def total_bank_amount(self):
        total_amount = 0
        for obj in Bank.instances:
            total_amount += obj.amount
        print(f"The total amount that bank has is {total_amount}")

    def calc_emi(self,principal_amount,years):
        
        self.emi = (principal_amount*0.07*(1 + 0.07)**(years*12)) / ((1 + 0.07)**(years*12) - 1)
        return self.emi

    def take_loan(self):
        if not self.is_login:
            print("Please login first")
            return 
        if self.has_loan:
            print("You already have a loan")
            return
            
        print("Loan amount must not be greater than 10 lakhs")
        amount = int(input("How much amount would you like? "))
        if amount > 1000000:
            print("Amount can't be greater than 10 lakhs")
            return

        if Bank.bank_amount < amount:
            print("Please try again later!")
            return

        
        print("Years of loan should not be more than 25")
        years = int(input("Enter the number of years: "))
        if years > 25:
            print("Years can't be greater than 25")
            return

        
        print(f"Your calcuated emi is {self.calc_emi(amount,years)}")

        
        confirm_loan = input("Would you like to confirm your loan? y/n : ")
        if confirm_loan == "y" or confirm_loan == "Y":
            self.amount += amount
            Bank.bank_amount -= amount
            print("Your loan has been confirmed")
            self.has_loan = True
            transaction = {
                "amount" : amount,
                "type" : "loan",
                "from" : "Bank",
                "to" : self.name
            }
            self.transaction.append(transaction)
        else:
            print("Your loan has been cancelled")

    def pay_emi(self):
        if not self.is_login:
            print("Please login first")
            return 
        if not self.has_loan:
            print("You don't have a loan")
            return
        if self.amount < self.emi:
            print("You don't have enough balance")
            return
        
        self.amount -= self.emi
        Bank.bank_amount += self.emi
        transaction = {
            "amount" : self.emi,
            "type" : "emi",
            "from" : self.name,
            "to" : "Bank"
        }
        self.transaction.append(transaction)
        print("You have successfully paid you emi")
        print(f"Your remaining balance is {self.amount}")

    def get_balance(self):
        self.get_account_dict()
        return f"Your balance is '{self.amount}'"
        
                
            

```

```python
try:

    account_dict = dict()
    user = Bank()
    while True:
        
        user_input = int(input("What would you like to do?\n1) Login\n2) Logout\n3) Deposit\n4) Withdraw\n5) Take loan\n6) Pay emi\n7) Close account\n8) Transfer amount\n9) Get Transactions\n10) Total bank Amount\n11) Exit\n12) Get balance\n13) Create account\n"))
        match user_input:
            case 1:
                user.login()
            case 2: 
                user.logout()
            case 3:
                amount = int(input("Enter The amount : "))
                user.deposit(amount)
            case 4:
                amount = int(input("Enter The amount : "))
                user.withdraw(amount)
            case 5:
                user.take_loan()
            case 6:
                user.pay_emi()
            case 7:
                user.close_account()
            case 8:
                user.transfer_amount()
            case 9:
                user.get_transactions()
            case 10:
                user.total_bank_amount()
            case 11:
                break
            case 12:
                print(user.get_balance())
            case 13:
                user = Bank()
except Exception as e:
    print(e)
     # 4941 - divya
    # 4181 - ganpat
```
