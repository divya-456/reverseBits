```python
class LessThanZeroException(Exception):
    def __init__(self):
        super().__init__("Integer value less than or equal to zero!")

class EmptyStringException(Exception):
    def __init__(self):
        super().__init__("Your string is empty")

class ValueNotFoundException(Exception):
    def __init__(self):
        super().__init__("Value not found!")

class OutOfRangeException(Exception):
    def __init__(self):
        super().__init__("Enter the value in the range specified")
```

# 2) Given a CSV file with student names and scores, find the student with the highest score

```python
from collections import defaultdict
try:
    marks_student = defaultdict(list)
    file = open("student_marks.csv")
    file.readline()
    for line in file:
        student_list = [i.strip() for i in line.split(",")]
        marks_student[student_list[1]].append(student_list[0])
    marks = sorted(marks_student,reverse=True)
    print(f"Student : {marks_student[marks[0]]}, marks : {marks[0]}")
    file.close()
except Exception as e:
    print(e)
```

# 3) Implement a program that reads a text file and counts the number of words and lines in it

```python
try:
    with open("demo1.txt") as file1:
        line_count = 0
        word_count = 0
        for line in file1:
            if line:
                line_count += 1
                word_count += len(line.split())
        print(f"Line count: {line_count} Word count : {word_count}")
except Exception as e:
    print(e)
```

# 4) Create a function that takes a list of sentences and writes them to a new text file, each on a new line 

```python
try:
    n = int(input("How many sentences do you want ? "))
    if n<=0:
        raise LessThanZeroException
    list_of_sentences = []
    for i in range(n):
        sentence = input("Enter your sentence: ")
        if sentence.strip() == "":
            raise EmptyStringException
        list_of_sentences.append(sentence)
    with open("demo3.txt","a") as file1:
        for i in list_of_sentences:
            file1.write(i+"\n")

    with open("demo3.txt","r") as file2:
        print(file2.read())
            
except LessThanZeroException as e:
    print(e)
except EmptyStringException as e:
    print(e)
except Exception as e:
    print(e)
```

# 5) Given a CSV file with employee details (name, age, salary), calculate the average salary of all employees

```python
try:
    with open("Employee_Salary_Dataset.csv") as file:
        file.readline()
        salary = []
        for line in file:
            salary.append(int(line.split(",")[2].strip()))
        sums = 0
        for i in salary:
            sums += i
        average = sums / len(salary)
        print(f"{average:.2f}")
except Exception as e:
    print(e)
```

# 6) Write a program that reads a CSV file and finds the total sales revenue for a specific product

```python
try:
    with open("SalesDataset.csv") as file:
        file.readline()
        product_dict = defaultdict(list)
        for line in file:
            product_dict[line.split(",")[0]].append(int(line.split(",")[1]))
        for i in product_dict.keys():
            print(i,end = " , ")
        product = input("For which product would you like to see the sales for? ")
        if product not in product_dict.keys():
            raise ValueNotFoundException
        if product.strip() == "":
            raise EmptyStringException
        sales = product_dict[product]
        sums = 0
        for i in sales:
            sums += i
        print(sums)
except ValueNotFoundException as e:
    print(e)
except EmptyStringException as e:
    print(e)
except Exception as e:
    print(e)
```

# 7) Given a text file with a list of numbers, write a function that finds the sum of all numbers in the file

```python
def find_sum(nums):
    sums = 0
    for i in nums:
        sums += i
    return sums
try:
    with open("numbers.txt") as file:
        nums =[int(i) for i in file.readline().split(",")]
        print(find_sum(nums))
except Exception as e:
    print(e)
```

# 9) Write a function that reads a JSON file and extracts specific information from it

```python
import json
try:
    def get_names(file):
        for index,student in enumerate(file):
            print(f"{index} - {student['name']} ")
    def get_marks(subject,file):
        for student in file:
            print(f"{student['name']} - {student[subject]}")

    with open("student.json") as file:
        file = json.load(file)
        info = int(input("Enter what would you like to know of our students :\n1) name of all student\n2) physics marks of all student\n3) maths marks of all student\n4) english marks of all student:\n"))
        if info <1 or info>4:
            raise OutOfRangeException
        match info:
            case 1:
                get_names(file)
            case 2:
                get_marks("physics",file)
            case 3:
                get_marks("maths",file)
            case 4:
                get_marks("english",file)
except OutOfRangeException as e:
    print(e)
except Exception as e:
    print(e)
```
