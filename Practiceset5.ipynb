```python
for i in range(n):
    for j in range(i,n):
        print(" ",end="")
    for k in range(i+1):
        print("*",end="")
    print()
```

```python
import collections

dict1 = { 'a' : 1, 'b' : 2 } 
dict2 = { 'b' : 3, 'c' : 4 } 

chain = collections.ChainMap(dict2, dict1) 
print ("All the ChainMap contents are : ") 
print (chain.maps) 
  

print ("All keys of ChainMap are : ") 
print (list(chain.keys())) 

print ("All values of ChainMap are : ") 
print (list(chain.values())) 

chain['b']
```
