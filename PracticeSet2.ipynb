# 2) Write a program that finds the largest and smallest elements in a list


```python
list1 = list(map(int,input("Enter numbers seperated by spaces").split()))
minimum, maximum = list1[0],0
for i in list1:
    if maximum < i:
        maximum = i
    if minimum > i:
        minimum = i
print(f"min: {minimum} , max: {maximum}")
```

# 3) Implement a function that takes a list of numbers and returns a new list with the squared values


```python
def squares(input_list):
    output_list = [i**2 for i in input_list]
    return output_list
input_list = list(map(int,input("Enter numbers seperated by spaces").split()))
print(squares(input_list))
```

```python
# second
def sort_strings(strings):
    for i in range(len(strings)):
        for j in range(len(strings)-i-1):
            if len(strings[j]) > len(strings[j+1]):
                strings[j], strings[j+1] = strings[j+1], strings[j]
    return strings
strings = input("Enter words seperated by space").split()
sort_strings(strings)
```

# 11) Given two dictionaries, merge them into a single dictionary 

```python
dict1 = {}
dict2 = {}
n1 = int(input("How many pairs do you want for your first dictionary? "))
i=0
while i<n1:
    key = input("Enter your key: ").strip()
    if not dict1.get(key):
        value = input("Enter your value: ").strip()
        dict1[key] = value
        i+=1
    else:
        print("Key already exists")
        
n2 = int(input("How many pairs do you want for your second dictionary? "))
j=0
while j<n2:
    key = input("Enter your key: ").strip()
    if not dict2.get(key):
        value = input("Enter your value: ").strip()
        dict2[key] = value
        j+=1
    else:
        print("Key already exists")
for key, value in dict2.items():
    if not dict1.get(key):
        dict1[key] = value
print(dict1)
```

# 12) Write a program that finds the most frequent element in a list
 

```python
def maximum(dict_values):
    maxim = 0
    for i in dict_values:
        if maxim < i:
            maxim = i
    return maxim


list1 = input("Enter element seperated by spaces: ").split()
frequency_dict = defaultdict(int)
for i in list1:
    frequency_dict[i] += 1

value = maximum(frequency_dict.values())
keys = list(filter(lambda key:frequency_dict[key] == value ,frequency_dict))
print(keys,value)
```

# 18) Create a function that takes a list of dictionaries and sorts them based on a specified key

# 19) Write a program that finds the average value of all the elements in a list of dictionaries

```python
list_of_dict = []
no_of_dict = int(input("How many dictionaries do you want? "))
for i in range(no_of_dict):
    single_dict = defaultdict(int)
    no_of_pairs = int(input(f"How many pairs do you want for your dict number {i+1}? "))
    
    for i in range(1,no_of_pairs+1):
        value = int(input("Enter a value"))
        single_dict[i] = value
    list_of_dict.append(single_dict)
count = 0
value_sum = 0
for dictionary in list_of_dict:
    for value in dictionary.values():
        value_sum += value
        count += 1
print(f"average : {value_sum/count:.2f}")
```

# 20) Implement a function that takes a list of strings and returns a set of unique characters in all strings.

```python
list_of_strings = []
n = int(input("Enter number of strings you want to enter: "))
for i in range(n):
    string = input("Enter your string: ").strip()
    list_of_strings.append(string)
# print(list_of_strings)
unique_char_set = set()
for sentence in list_of_strings:
    character_dict = defaultdict(int)
    string = ''
    for character in sentence:
        if not character_dict.get(character):
            character_dict[character] += 1
    for key in character_dict.keys():
        string += key
    unique_char_set.add(string)
print(unique_char_set)
```
