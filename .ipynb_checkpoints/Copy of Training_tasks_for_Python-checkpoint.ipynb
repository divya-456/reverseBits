## W.A.P. that replicates a calculator / Convert this to function/class based script when you learn about them

```python
# create naive script here

try:

  num1 = int(input("Enter first number: "))
  num2 = int(input("Enter second number: "))
  operator = input("Enter operator: ")

  if operator == "+":
    print(num1 + num2)
  elif operator == "-":
    print(num1 - num2)
  elif operator == "*":
    print(num1 * num2)
  elif operator == "/":
    print(num1 / num2)

except ValueError:
  print("Invalid input")
except ZeroDivisionError:
  print("Cannot divide by zero")
except Exception as e:
  print(e)
```

```python
# create function based script here

def sum(num1,num2):
  return num1 + num2

def subtract(num1,num2):
  return num1 - num2

def multiply(num1,num2):
  return num1 * num2

def divide(num1,num2):
  return num1 / num2

try:

  num1 = int(input("Enter first number: "))
  num2 = int(input("Enter second number: "))
  operator = input("Enter operator: ")

  if operator == "+":
    print(sum(num1,num2))
  elif operator == "-":
    print(subtract(num1,num2))
  elif operator == "*":
    print(multiply(num1,num2))
  elif operator == "/":
    print(divide(num1,num2))

except ValueError:
  print("Invalid input")
except ZeroDivisionError:
  print("Cannot divide by zero")
except Exception as e:
  print(e)
```

```python
# create class based script here

class Calculator:
  def __init__(self,num1,num2):
    self.num1 = num1
    self.num2 = num2

  def sum(self):
    return self.num1 + self.num2

  def subtract(self):
    return self.num1 - self.num2

  def multiply(self):
    return self.num1 * self.num2

  def divide(self):
    return self.num1 / self.num2

try:

  num1 = int(input("Enter first number: "))
  num2 = int(input("Enter second number: "))
  calc = Calculator(num1,num2)
  operator = input("Enter operator: ")

  if operator == "+":
    print(calc.sum())
  elif operator == "-":
    print(calc.subtract())
  elif operator == "*":
    print(calc.multiply())
  elif operator == "/":
    print(calc.divide())


except ValueError:
  print("Invalid input")
except ZeroDivisionError:
  print("Cannot divide by zero")
except Exception as e:
  print(e)

```

## W.A.P. that calculates result of a student. Student has given exam for five subjects. Passing marks are 23 out of 70


```python
# create naive script here
try:
  name = input("Enter name: ")
  marks = []

  for j in range(5):
    marks.append(int(input("Enter marks: ")))

  total = sum(marks)
  percentage = (total / 350)*70

  if percentage >= 23:
    print(f"{name} - {percentage}% - passed")
  else:
    print(f"{name} - {percentage}% - failed")

except ValueError as e:
  print(e)
except Exception as e:
  print(e)
```

```python
# create function based script here

def result(marks):
  total = sum(marks)
  percentage = (total / 350)*70

  if percentage >= 23:
    print(f"{name} - {percentage}% - passed")
  else:
    print(f"{name} - {percentage}% - failed")

try:
  name = input("Enter name: ")
  marks = []

  for j in range(5):
    marks.append(int(input("Enter marks: ")))

  result(marks)
except ValueError as e:
  print(e)

except Exception as e:
  print(e)
```

```python
# create class based script here
class Result:
  def __init__(self,name,marks):
    self.name = name
    self.marks = marks

  def calc_result(self):
    total = sum(self.marks)
    percentage = (total / 350)*70

    if percentage >= 23:
      print(f"{self.name} - {percentage}% - passed")
    else:
      print(f"{self.name} - {percentage}% - failed")

try:
  name = input("Enter name: ")
  marks = []

  for j in range(5):
    marks.append(int(input("Enter marks: ")))

  result = Result(name,marks)
  result.calc_result()
except ValueError as e:
  print(e)

except Exception as e:
  print(e)
```
