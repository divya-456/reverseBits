# 3) Implement a program that simulates a basic bank account using a BankAccount class.


```python
class Bank:
    def __init__(self,name,acc_no,pin,initial_amount):
        self.name = name
        self.acc_no = acc_no
        self._pin = pin
        self.amount = initial_amount
        
    @property
    def pin(self):
        return "Sorry we can't show you the pin"

    @pin.setter
    def pin(self,new_pin):
        try:
            acc = int(input("Please enter your account number"))
            if acc != self.acc_no:
                return "Please enter valid account number!"
            self._pin = new_pin
        except Exception as e:
            print(e)

    @property
    def get_details(self):
        return f"Your name: {self.name} , Your Account number : {self.acc_no}, Your current amount : {self.amount}"

    def withdraw(self,amount):
        try:
            acc = int(input("Please enter your account number"))
            if acc == self.acc_no:
                name = input("Please enter your name")
                if name == self.name:
                    pin = int(input("Please enter your pin"))
                    if pin == self._pin:
                        if checkbalance(amount):
                            self.amount -= amount
                            print(f"Your current balance is {self.amount}")
                        return "Not enough balance"
                    return "Please enter correct pin"
                return "Please enter correct name"
            return "Please enter valid account number"
        except Exception as e:
            print(e)
                        
        
    def deposit(self,amount):
        try:
            acc = int(input("Please enter your account number"))
            if acc == self.acc_no:
                self.amount += amount
                print(f"Your current balance is {self.amount}")
            return "Please enter a valid account number"
        except Exception as e:
            print(e)

    def checkbalance(self,amount):
        if self.amount - amount > 2000:
            return True
        return False
```
