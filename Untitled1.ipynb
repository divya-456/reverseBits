```python
from functools import total_ordering

@total_ordering
class Account:
    def __init__(self,owner,amount=0):
        self.owner = owner
        self.amount = amount
        self._transactions = []

    def __str__(self):
        return f"{self.owner} ,{self.amount}"

    def __repr__(self):
        return f"Repr - {self.owner} , {self.amount}"

    def add_transactions(self,amount):
        self._transactions.append(amount)

    @property
    def get_balance(self):
        return self.amount + sum(self._transactions)

    def __len__(self):
        return len(self._transactions)

    def __getitem__(self,position):
        return self._transactions[position]
    
    def __reversed__(self):
        return self[::-1]

    def __eq__(self,other):
        return self.get_balance == other.get_balance

    def __lt__(self,other):
        return self.get_balance < other.get_balance

    def __add__(self,other):
        owner = f"{self.owner}&{other.owner}"
        start_amount = self.amount + other.amount
        acc = Account(owner,start_amount)
        for t in list(self._transactions) + list(other._transactions):
            acc.add_transaction(t)
        return acc

    def __enter__(self):
        print("Making a copy of transaction:")
        self._copy_transaction = self._transactions
        print("Copy successfull")

    def __exit__(self,exc_type,exc_value,exc_tb):
        print("checking for exception: ")
        if exc_type:
            print("Rolling back to previous transaction!")
            self._transactions = self._copy_transaction
            print(f"Transaction resulted in {exc_type} : {exc_value}")




    
```

```python
def validate_transaction(acc,amount):
    with acc as a:
        print("Adding the amount")
        a.add_transactions(amount)
        if a.get_balance < 0:
            raise ValueError("Sorry can't go in debt!")
                
        
```
