# 5) Create a Python function to check if a given string is a palindrome

```python
def is_pallindrome(string:str)-> bool:
    reverse = ''
    for i in range(len(string)-1,-1,-1):
        reverse +=string[i]
    if string == reverse:
        return True
    return False
string = input("Enter a word: ")
if(is_pallindrome(string)):
    print(f"{string} is a pallindrome")
else:
    print(f"{string} is not a pallindrome")

#TODO: 
```

# 19) Create a function to count the number of vowels in a given string

# 24) Create a program that generates the Fibonacci sequence up to a given number of terms)


```python
def fibonacci(n):
    if n<=1:
        return n
    else:
        return (fibonacci(n-1) + fibonacci(n-2))
n = int(input("Enter a number: "))
if n<=0:
    print("Enter a valid number")
else:
    
    for i in range(n):
        print(fibonacci(i),end=",")
```

# 27) Write a program that calculates the factorial of a given number

```python
def fact(n):
    if n==0 or n == 1:
        return 1
    else:
        return n*fact(n-1)
num = int(input("Enter a number: "))
fact(num)
```

# 28) Create a loop that prints all prime numbers between 1 and 50

```python
def is_prime(num):
    if num == 1:
        return False
    for i in range(2,math.floor(math.sqrt(num))+1):
        if num%i == 0:
            return False
    return True

for j in range(1,51):
    if(is_prime(j)):
        print(j)
```
